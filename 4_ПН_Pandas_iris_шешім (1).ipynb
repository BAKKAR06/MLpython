## Топтау

**Мақсаты:** гүлдерді түрлері бойынша топтап, әр топтың жиынтық көрсеткіштерін есептеу.

### Тапсырма 17. Түрлер бойынша орташа

Әр түр бойынша төрт өлшемнің (`sepal_length`, `sepal_width`, `petal_length`, `petal_width`) орташасын екі ондық таңбаға дейін есептеңіз. Қай өлшем бойынша түрлер бір-бірінен ең қатты ерекшеленетінін комментарийде жазыңыз.

**Күтілетін нәтиже:**

```
            sepal_length  sepal_width  petal_length  petal_width
species                                                         
setosa              5.01         3.43          1.46         0.25
versicolor          5.94         2.77          4.26         1.33
virginica           6.59         2.97          5.55         2.03
```

````python
FEATURES = ["sepal_length", "sepal_width", "petal_length", "petal_width"]

print(iris.groupby("species")[FEATURES].mean().round(2))

# Түрлер бір-бірінен petal_length (күлте жапырақтың ұзындығы) бойынша ең қатты ерекшеленеді:
# setosa - 1.46, versicolor - 4.26, virginica - 5.55 см (айырмашылық 4 см-ге жуық).
# Ал sepal_width бойынша айырмашылық ең аз (2.77 - 3.43 см).
````

### Тапсырма 18. Атаулы агрегация

`agg()` арқылы әр түр үшін жиынтық кесте құрыңыз: `count` — гүлдер саны, `pl_mean` — `petal_length` орташасы, `sl_max` — `sepal_length` ең үлкен мәні, `sw_std` — `sepal_width` стандартты ауытқуы. Нәтижені екі ондық таңбаға дейін дөңгелектеңіз.

**Күтілетін нәтиже:**

```
            count  pl_mean  sl_max  sw_std
species                                   
setosa         50     1.46     5.8    0.38
versicolor     50     4.26     7.0    0.31
virginica      50     5.55     7.9    0.32
```

````python
summary = iris.groupby("species").agg(
    count=("species", "size"),
    pl_mean=("petal_length", "mean"),
    sl_max=("sepal_length", "max"),
    sw_std=("sepal_width", "std"),
).round(2)

print(summary)
````

### Тапсырма 19. transform

`transform()` арқылы әр гүлдің жанына өз түрінің `sepal_length` орташасын көрсететін `species_mean_sl` бағанын қосыңыз. Тостағанша жапырағының ұзындығы өз түрінің орташасынан асатын гүлдерді таңдап, олардың әр түр бойынша санын шығарыңыз.

**Күтілетін нәтиже:**

```
species
setosa        22
versicolor    24
virginica     22
dtype: int64
```

````python
iris["species_mean_sl"] = iris.groupby("species")["sepal_length"].transform("mean")

above = iris[iris["sepal_length"] > iris["species_mean_sl"]]
print(above.groupby("species").size())
````

### Тапсырма 20. Жиынтық кесте

`pd.cut()` функциясы арқылы `sepal_length` мәндерін төрт өлшем санатына бөліп, `size` бағанына жазыңыз: `bins=[4, 5, 6, 7, 8]`, `labels=["кіші", "орташа", "үлкен", "өте үлкен"]`. Аралықтар сол жақтан ашық, оң жақтан жабық: мысалы, 5.0 мәні "кіші" санатына түседі.

Содан кейін `pd.pivot_table()` арқылы жолдары түр, бағандары өлшем санаты болатын, әр ұяшықта гүлдер саны жазылған кесте құрыңыз (`aggfunc="count"`, `fill_value=0`, `observed=False`).

**Күтілетін нәтиже:**

```
size        кіші  орташа  үлкен  өте үлкен
species                                   
setosa        28      22      0          0
```

## Ең жақын орталық бойынша түрін анықтау

**Мақсаты:** топтауды, кеңейтуді, векторланған есептеулерді және фильтрлеуді бір толық есепте бірге қолдану.

Ботаник гүлдің түрін оның төрт өлшемі бойынша автоматты түрде анықтағысы келеді. Ұсынылған тәсіл қарапайым: әр түрдің "орталығын", яғни сол түрге жататын гүлдердің төрт өлшемінің орташасын есептейміз. Жаңа гүлдің түрі ретінде оған ең жақын орталықтың түрі алынады. Екі нүктенің арасындағы қашықтық евклидтік формуламен есептеледі:

`d = sqrt((x1 - c1)² + (x2 - c2)² + (x3 - c3)² + (x4 - c4)²)`

Тапсырма LeetCode стилінде 4 бөлімге бөлінген. Әр бөлім алдыңғы бөлімде жазылған функцияның нәтижесін қолданады. Мысалдар бес гүлден тұратын шағын `toy` кестесінде берілген, ал күтілетін нәтиже толық `iris` кестесі бойынша есептелген.

**Шектеулер:**

- өлшемдер тізімі: `FEATURES = ["sepal_length", "sepal_width", "petal_length", "petal_width"]`;
- кестенің жолдары бойынша цикл жазуға болмайды; түрлер бойынша (3 қадам) цикл жазуға рұқсат етіледі;
- функциялар бастапқы кестені өзгертпеуі керек.

**Мысалдарға арналған шағын кесте:**

```python
toy = pd.DataFrame({
    "sepal_length": [5.0, 5.2, 6.8, 7.0, 5.9],
    "sepal_width": [3.4, 3.6, 3.0, 3.2, 3.1],
    "petal_length": [1.5, 1.4, 5.6, 5.8, 4.9],
    "petal_width": [0.2, 0.3, 2.1, 2.3, 1.7],
    "species": ["setosa", "setosa", "virginica", "virginica", "setosa"],
})
```

Соңғы гүл (индексі 4) әдейі "қате белгіленген": өлшемдері `virginica` түріне ұқсас, бірақ `setosa` деп жазылған.

### Бөлім 1. Түрлердің орталығы

**Функция:** `centroids(df)`

Әр түр бойынша төрт өлшемнің орташасын есептеп, жолдары түр атауы, бағандары `FEATURES` болатын DataFrame қайтарыңыз.

**Мысал:**

```
Кіріс:  centroids(toy)
Шығыс:
           sepal_length  sepal_width  petal_length  petal_width
species                                                        
setosa            5.367        3.367           2.6        0.733
virginica         6.900        3.100           5.7        2.200
```

**Күтілетін нәтиже (`centroids(iris).round(3)`):**

```
            sepal_length  sepal_width  petal_length  petal_width
species                                                         
setosa             5.006        3.428         1.462        0.246
versicolor         5.936        2.770         4.260        1.326
virginica          6.588        2.974         5.552        2.026
```

````python
FEATURES = ["sepal_length", "sepal_width", "petal_length", "petal_width"]

toy = pd.DataFrame({
    "sepal_length": [5.0, 5.2, 6.8, 7.0, 5.9],
    "sepal_width": [3.4, 3.6, 3.0, 3.2, 3.1],
    "petal_length": [1.5, 1.4, 5.6, 5.8, 4.9],
    "petal_width": [0.2, 0.3, 2.1, 2.3, 1.7],
    "species": ["setosa", "setosa", "virginica", "virginica", "setosa"],
})


def centroids(df):
    """Әр түр бойынша төрт өлшемнің орташасы."""
    return df.groupby("species")[FEATURES].mean()


print(centroids(toy).round(3))
print()

C = centroids(iris)
print(C.round(3))
````

### Бөлім 2. Орталықтарға дейінгі қашықтық

**Функция:** `distances(df, C)`

`C` — 1-бөлімнің нәтижесі. Әр гүлден әр түрдің орталығына дейінгі қашықтықты есептеп, жолдары гүлдер (`df` индексі), бағандары түр атаулары болатын DataFrame қайтарыңыз.

**Кеңес:** `df[FEATURES] - C.loc["setosa"]` өрнегінде Series баған атаулары бойынша кестенің әр жолынан алынады (кеңейту). Түрлер бойынша цикл жазып, нәтижелерді сөздікке жинаңыз да, одан DataFrame құрыңыз.

**Мысал:**

```
Кіріс:  distances(toy, centroids(toy)).round(3)
Шығыс:
   setosa  virginica
0   1.277      5.034
1   1.308      5.024
2   3.613      0.200
3   3.923      0.200
4   2.565      1.375
```

**Күтілетін нәтиже (`distances(iris, C)` нәтижесінің алғашқы 3 жолы және пішіні):**

```
   setosa  versicolor  virginica
0   0.141       3.268      4.803
1   0.448       3.252      4.840
2   0.417       3.427      5.003
(150, 3)
```

````python
def distances(df, C):
    """Әр гүлден әр түрдің орталығына дейінгі евклидтік қашықтық."""
    result = {}
    for sp in C.index:                       # түрлер бойынша цикл (3 қадам)
        diff = df[FEATURES] - C.loc[sp]      # кеңейту: әр жолдан орталық алынады
        result[sp] = np.sqrt((diff ** 2).sum(axis=1))
    return pd.DataFrame(result, index=df.index)


print(distances(toy, centroids(toy)).round(3))
print()

D = distances(iris, C)
print(D.head(3).round(3))
print(D.shape)
````

### Бөлім 3. Түрді анықтау

**Функция:** `predict(D)`

`D` — 2-бөлімнің нәтижесі. Әр жол үшін ең жақын орталықтың (ең кіші қашықтықтың) баған атауын Series ретінде қайтарыңыз.

**Кеңес:** `idxmin(axis=1)` әр жолдағы ең кіші мәннің баған атауын береді.

**Мысал:**

```
Кіріс:  predict(distances(toy, centroids(toy)))
Шығыс:
0       setosa
1       setosa
2    virginica
3    virginica
4    virginica
dtype: str
```

**Күтілетін нәтиже (`predict(D).value_counts()`):**

```
versicolor    53
setosa        50
virginica     47
Name: count, dtype: int64
```

````python
def predict(D):
    """Әр жол үшін ең жақын орталықтың түрі."""
    return D.idxmin(axis=1)


print(predict(distances(toy, centroids(toy))))
print()

pred = predict(D)
print(pred.value_counts())
````

### Бөлім 4. Нәтижені бағалау

**Функция:** `evaluate(df, pred)`

Кортеж қайтарыңыз:

- дұрыс анықталған гүлдердің үлесі (төрт ондық таңбаға дейін);
- жолдары нақты түр, бағандары анықталған түр болатын `pd.crosstab` кестесі (`rownames=["нақты"]`, `colnames=["болжам"]`);
- түрі қате анықталған гүлдердің индекстер тізімі.

**Мысал:**

```
Кіріс:  evaluate(toy, predict(distances(toy, centroids(toy))))
Шығыс:
0.8
болжам     setosa  virginica
нақты                       
setosa          2          1
virginica       0          2
[4]
```

**Күтілетін нәтиже (`iris` бойынша):**

```
0.9267
болжам      setosa  versicolor  virginica
нақты                                    
setosa          50           0          0
versicolor       0          46          4
virginica        0           7         43
[50, 52, 76, 77, 106, 113, 119, 121, 126, 127, 138]
```

**Қосымша сұрақ (комментарийде жауап беріңіз):** қай түр басқа түрлермен ешқашан шатастырылмады және оның себебі қандай? Жауап беру үшін 17-тапсырмадағы түрлер бойынша орташа мәндерді салыстырыңыз.

````python
def evaluate(df, pred):
    """Дәлдік, шатасу кестесі және қате анықталған гүлдердің индекстері."""
    correct = df["species"] == pred
    accuracy = round(correct.mean(), 4)
    table = pd.crosstab(df["species"], pred, rownames=["нақты"], colnames=["болжам"])
    wrong = df.index[~correct].tolist()
    return accuracy, table, wrong


acc, ct, wrong = evaluate(toy, predict(distances(toy, centroids(toy))))
print(acc)
print(ct)
print(wrong)
print()

acc, ct, wrong = evaluate(iris, pred)
print(acc)
print(ct)
print(wrong)

# Қосымша сұрақтың жауабы:
# setosa түрі басқа түрлермен ешқашан шатастырылмады (50 гүлдің 50-і дұрыс).
# Себебі 17-тапсырмадағы орташа мәндер бойынша setosa күлте жапырағы өте кішкентай:
# petal_length = 1.46, petal_width = 0.25 см, ал versicolor (4.26 / 1.33) мен
# virginica (5.55 / 2.03) әлдеқайда үлкен. Сондықтан setosa орталығы басқа екі
# орталықтан алыс жатыр. Ал versicolor мен virginica өлшемдері бір-біріне жақын,
# сондықтан қателердің бәрі (4 + 7 = 11) осы екі түрдің арасында болды.
````
